```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
```

```python
fig, ax = plt.subplots()
ax.scatter(1, 2, zorder=3, color='black')

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-0.5, 2.5)

ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
ax.yaxis.set_major_locator(ticker.MultipleLocator(1))

# hide the "0" label on both axes
def hide_zero(x, pos):
    return '' if x == 0 else f'{int(x)}'

ax.xaxis.set_major_formatter(ticker.FuncFormatter(hide_zero))
ax.yaxis.set_major_formatter(ticker.FuncFormatter(hide_zero))

ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.tick_params(axis='x', pad=10)
ax.tick_params(axis='y', pad=10)

# equal spacing between ticks on both axes
ax.set_aspect('equal', adjustable='box')

fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

fig.savefig('plot.png', transparent=True, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

fig, ax = plt.subplots()

p1 = (1, 2)
p2 = (5, 3)

ax.scatter(*p1, zorder=3, color='black')
ax.scatter(*p2, zorder=3, color='black')

ax.annotate('', xy=p2, xytext=p1,
            arrowprops=dict(arrowstyle='->', color='red', lw=2),
            zorder=2)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_xlim(-0.5, 5.5)
ax.set_ylim(-0.5, 3.5)

ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
ax.yaxis.set_major_locator(ticker.MultipleLocator(1))

def hide_zero(x, pos):
    return '' if x == 0 else f'{int(x)}'

ax.xaxis.set_major_formatter(ticker.FuncFormatter(hide_zero))
ax.yaxis.set_major_formatter(ticker.FuncFormatter(hide_zero))

ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')
ax.tick_params(axis='x', pad=10)
ax.tick_params(axis='y', pad=10)

ax.set_aspect('equal', adjustable='box')

fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

fig.savefig('plot2.png', transparent=True, bbox_inches='tight')
plt.show()
```
